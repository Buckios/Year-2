```python
T = np.array([[0.6, 0.7],
              [0.4, 0.3]])

p = np.array([[0.5],
              [0.5]])

k_values = [1, 2, 10, 100, 100000]

for k in k_values:
    pk = np.linalg.matrix_power(T, k) @ p
    print(f"k = {k}:\n{pk}\n")
```
