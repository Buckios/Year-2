```python
import numpy as np
```

```python
matrix = np.random.randint(1, 10, size=(100, 100))

result = np.where(matrix > 5, 1, np.where(matrix < 5, 0, matrix))

print("Input: \n", matrix)
print("Output: \n", result, "\n")
```

```python
matrix = np.random.randint(1, 10, size=(100, 100))

sum = np.trace(matrix)

print("Input: \n", matrix)
print("Sum of the main diagonal elements of the matrix:", sum, "\n")
```

```python
n = 15
magic_square = np.zeros((n, n), dtype=int)

i, j = 0, n // 2

for num in range(1, n**2 + 1):
    magic_square[i, j] = num
    next_i, next_j = (i - 1) % n, (j + 1) % n
    
    if magic_square[next_i, next_j] != 0:
        i = (i + 1) % n
    else:
        i, j = next_i, next_j

print("15x15 Magic Square:\n", magic_square, "\n")
```

```python
matrix = np.ones((100, 100), dtype=int)

matrix[1:-1, 1:-1] = 0

print("Matrix with 1 on borders and 0 inside:\n", matrix, "\n")
```

```python
matrix = np.random.randint(1, 10, size=(100, 100))

border_matrix = np.pad(matrix, pad_width=1, mode='constant', constant_values=0)

print("Input: \n", matrix)  
print("Matrix with 0 border:\n", border_matrix, "\n")
```

```python
matrix = np.random.randint(1, 10, size=(100, 100))

sort_matrix = np.sort(matrix, axis=1)

print("Input \n", matrix)
print("Matrix with sorted rows:\n", sort_matrix, "\n")
```

```python
matrix = np.random.randint(1, 101, size=(100, 100))

multiples_sum = matrix[(matrix % 3 == 0) | (matrix % 5 == 0)].sum()

print("Input: \n", matrix)
print("Sum of all multiples of 3 or 5:", multiples_sum, "\n")
```

```python
matrix = np.random.randint(1, 1001, size=(100, 100))

five_smallest = np.sort(matrix, axis=None)[:5]

smallest_list = five_smallest.tolist()

print("Input: \n", matrix)
print("Five smallest elements:", smallest_list, "\n")
```
